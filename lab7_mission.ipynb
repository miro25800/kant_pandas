```python
# Lab 7 · 미션: 내 플레이리스트 분석 (엔터 · 음악 스트리밍)

# 1부: 표 만들기 · 총합 · 조건 필터 · Top 3 (7-1 ~ 7-4)

import pandas as pd

# 7-1. 플레이리스트 DataFrame 생성
playlist = pd.DataFrame({
    "title": ["봄바람", "Run Up", "새벽 감성", "Neon City", "첫눈", "Jump!", "바다 소리", "Fire Dance"],
    "genre": ["발라드", "댄스", "발라드", "힙합", "발라드", "댄스", "인디", "댄스"],
    "plays": [12, 35, 8, 22, 15, 28, 5, 19],
})
playlist
```

```python
# 7-2. 이번 달 총 재생 수 (144)
total_plays = playlist["plays"].sum()
print("이번 달 총 재생 수:", total_plays)
```

```python
# 7-3. 재생 수 20 이상인 곡
over_20 = playlist[playlist["plays"] >= 20]
print("\n[재생 수 20 이상]")
print(over_20)
```

```python
# 7-4. 가장 많이 들은 Top 3 (내림차순 정렬 후 상위 3개)
top_3 = playlist.sort_values("plays", ascending=False).head(3)
print("\n[Top 3]")
print(top_3)
```

```python
# 2부: 장르별 묶기 · 복합 조건 필터링 (7-5 ~ 7-6)

# 7-5. 장르별 재생 수 합계 (큰 순 → 댄스 82로 1위)
genre_plays = playlist.groupby("genre")["plays"].sum().sort_values(ascending=False)
print("[장르별 재생 수 합계]")
print(genre_plays)
```

```python
# 7-6. 보너스: 발라드이면서 10번 이상 들은 곡 (각 조건 괄호 필수)
ballad_10_plus = playlist[(playlist["genre"] == "발라드") & (playlist["plays"] >= 10)]
print("\n[발라드 + 10회 이상]")
print(ballad_10_plus)
```
