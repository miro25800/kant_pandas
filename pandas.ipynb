```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

print(amounts)
print("합계", sum(amounts))

print(amounts * 2)
print("==="*3)
```

```python
import numpy as np

arr = np.array(amounts)
arr * 2

print("합계:", arr.sum())
print("평균:", round(arr.mean()))
print("최소:", arr.min())
print("최고:", arr.max())

print("모양:", arr.shape, " ,데이터 타입: ", arr.dtype)
```

```python
# 중요한 개념이라고 함.
# arr: NDArray[Any]

print(arr >= 10_000)
arr[arr >= 10000]
```

```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

print(shops)
# print("식비:", arr.shape, " ,데이터타입: ", arr.dtype) / 내가 만든 잘못 생각한 식비 합치기에 대한 코드
food = 0
for c, a in zip(categories, amounts):
    if c == "식비":
        food = food + a

print(food)
```

```python
import pandas as pd

spend = pd.DataFrame({
    "day": [2, 3, 4, 5, 7, 9, 12, 14, 16, 19, 20, 25],
    "shop": shops,
    "category": categories,
    "amount": amounts
})
spend.head(5)
spend.info()
# print(spend.shape)
```

```python
spend["amount"]
print("합계:", spend["amount"].sum())
print("평균:", round(spend["amount"].mean(),2))
```

```python
# spend["shop", "amounts"] "shop, amount"
# spend[ 컬럼명 ]
# spend[ [리스트] ] -> spend[ [컬럼1, 컬럼2, ...] ]
spend[["shop", "amount"]]
```

### 조건 설정

```python
spend["category"] == "문화"
```

```python
# 굉장히 축약적이나, 난해한 것이라 함.
# 문자열로 들어갔을때 행으로 들어갔을때 모두 틀리다고 함.

# spend[ 조건 ] 조건이 True인 행만 리턴
# sped[ "컬럼1"] -> "컬럼1"의 로우(row) 모두 리턴

spend[spend["category"] == "문화"]
```

```python
# 조건식
# spend[ 조건식 ] -> 조건식을 만족하는 (True) 모든 행(row)를 가져온다
# spend["amount"] >= 10_000
spend[spend["amount"] >= 10_000]
```

```python
# 엔드 조건
# 식비 5000원 이상 사용한 행을 출력
spend[(spend["category"] == "식비") & (spend["amount"] >= 5000)]
```
